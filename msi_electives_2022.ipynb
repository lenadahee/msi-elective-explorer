**Basically checking if course is in each track's required list. If true, then the course is one of the track's required courses. Since some students could have taken a course for a different track as an elective, each course becomes a vote for the pathway. Whichever pathway gets the most votes will be the inferred pathway.**

```python
# Course Pathway Course List by Required Classes Only
track_required_courses = {
    "BDA": ["SI 507", "SI 544", "SI 618", "SI 671"],
    "UX": ["SI 520"],
    "UCAD": ["SI 579", "SI 664", "SI 669"],
    "LAKES": ["SI 510", "SI 547", "SI 666", "SI 667", "SI 676", "SI 580", "SI 647"]
}

# Drop old flag columns if rerunning
df_filtered = df_filtered.drop(
    columns=[c for c in df_filtered.columns if c.lower().endswith("_flag")],
    errors="ignore"
)

# Create flag columns FIRST
for track, courses in track_required_courses.items():
    df_filtered[f"{track}_Flag"] = df_filtered["Course_Code"].isin(courses).astype(int)

# Then aggregate
track_counts = df_filtered.groupby("Emplid").agg(
    BDA_Count=("BDA_Flag", "sum"),
    UX_Count=("UX_Flag", "sum"),
    LAKES_Count=("LAKES_Flag", "sum"),
    UCAD_Count=("UCAD_Flag", "sum")
)

track_counts.head()
```

**Summarize values for each flag, aka the number of "True"**

**Ex. Student 00391172 took 2 BDA required courses, 1 UX Required Courses,0 LAKES, and 0 UCAD. Will take the highest # and infer that the student is BDA.**

```python
# Find students with zero required courses for any track
zero_track_students = track_counts[
    (track_counts["BDA_Count"] == 0) &
    (track_counts["UX_Count"] == 0) &
    (track_counts["LAKES_Count"] == 0) &
    (track_counts["UCAD_Count"] == 0)
]

print(zero_track_students)

print(len(zero_track_students), "students have zero required courses for any track.")
```

```python
# Create indicator columns for whether each course is in the required courses for each track.
# Will later beummed per student to count how many required courses they took for each track.
for track, courses in track_required_courses.items():
    df_filtered[f"{track}_flag"] = df_filtered["Course_Code"].isin(courses)

# Use the counts of required courses ("votes") to infer the most likely track for each student.
# Calculate total votes, detect ties, and identify the track with the highest count.
vote_cols = ["BDA_Count", "UX_Count", "LAKES_Count", "UCAD_Count"]

# total votes + max vote per student
total_votes = track_counts[vote_cols].sum(axis=1)
max_vote = track_counts[vote_cols].max(axis=1)

# how many tracks tied for the max
number_at_max = track_counts[vote_cols].eq(max_vote, axis=0).sum(axis=1)

# best guess track
best_guess = (
    track_counts[vote_cols]
    .idxmax(axis=1)
    .str.replace("_Count", "", regex=False)
)
```

```python
# Initialize all students as "Unknown", then assign tracks based on vote results.
track_counts["Inferred_Track"] = "Unknown"

# 0 votes → Zero-track (Students with no required-course signals are labeled "Zero-track".)
track_counts.loc[total_votes.eq(0), "Inferred_Track"] = "Zero-track"

# If multiple tracks share the highest vote count, label as "Tie".
tie_mask = total_votes.gt(0) & number_at_max.gt(1)
track_counts.loc[tie_mask, "Inferred_Track"] = "Tie"
# non-ties → assign best guess
assign_mask = total_votes.gt(0) & number_at_max.eq(1)
track_counts.loc[assign_mask, "Inferred_Track"] = best_guess[assign_mask]

track_counts["Inferred_Track"].value_counts()
```

```python
tie_ids = track_counts[track_counts["Inferred_Track"] == "Tie"].index
print(track_counts.loc[tie_ids, ["BDA_Count", "UX_Count", "LAKES_Count", "UCAD_Count"]].value_counts())
```

```python
# Drop point columns if they already exist (makes cell safe to rerun)
point_cols = ["UX_Points", "UCAD_Points", "BDA_Points", "LAKES_Points"]
track_counts = track_counts.drop(columns=[c for c in point_cols if c in track_counts.columns])

# Comprehensive tiebreaker point system
UX_TIEBREAK    = ["SI 520", "SI 539", "SI 582", "SI 588", "SI 622"]
UCAD_TIEBREAK  = ["SI 579", "SI 664", "SI 669"]
BDA_TIEBREAK   = ["SI 608", "SI 630", "SI 649", "SI 650", "SI 670"]
LAKES_TIEBREAK = ["SI 623", "SI 530", "SI 632", "SI 633", "SI 638",
                  "SI 583", "SI 643", "SI 552", "SI 519"]

df_filtered["UX_Points"]    = df_filtered["Course_Code"].isin(UX_TIEBREAK).astype(int)
df_filtered["UCAD_Points"]  = df_filtered["Course_Code"].isin(UCAD_TIEBREAK).astype(int)
df_filtered["BDA_Points"]   = df_filtered["Course_Code"].isin(BDA_TIEBREAK).astype(int)
df_filtered["LAKES_Points"] = df_filtered["Course_Code"].isin(LAKES_TIEBREAK).astype(int)

points = df_filtered.groupby("Emplid")[
    ["UX_Points", "UCAD_Points", "BDA_Points", "LAKES_Points"]
].sum()

track_counts = track_counts.join(points, how="left")
track_counts[["UX_Points", "UCAD_Points", "BDA_Points", "LAKES_Points"]] = (
    track_counts[["UX_Points", "UCAD_Points", "BDA_Points", "LAKES_Points"]].fillna(0)
)
```

```python
# Reassign tie students using all four track signals
tie_mask = track_counts["Inferred_Track"].isin(["Tie", "Manual"])

point_cols = ["BDA_Points", "UX_Points", "LAKES_Points", "UCAD_Points"]
col_to_track = {
    "BDA_Points":   "BDA",
    "UX_Points":    "UX",
    "LAKES_Points": "LAKES",
    "UCAD_Points":  "UCAD"
}

for idx in track_counts[tie_mask].index:
    row = track_counts.loc[idx, point_cols]
    max_pts = row.max()
    if max_pts == 0 or (row == max_pts).sum() > 1:
        track_counts.loc[idx, "Inferred_Track"] = "Tie"
    else:
        track_counts.loc[idx, "Inferred_Track"] = col_to_track[row.idxmax()]

track_counts["Inferred_Track"].value_counts()
```

## **ELECTIVE ANALYSIS**

```python
# Re-merge finalized track labels into course dataset
track_labels = track_counts.reset_index()[["Emplid", "Inferred_Track"]]
df_with_tracks = df_filtered.merge(track_labels, on="Emplid", how="left")

print(df_with_tracks["Inferred_Track"].value_counts())
print(f"Unique students: {df_with_tracks['Emplid'].nunique()}")  # should be 200
```

```python
# Required courses per track + universal exclusions
REQUIRED = {
    "BDA":   {"SI 507", "SI 544", "SI 618", "SI 671"},
    "UX":    {"SI 520", "SI 539", "SI 582", "SI 588", "SI 622"},  # added shared prereqs
    "UCAD":  {"SI 579", "SI 664", "SI 669", "SI 539", 
              "SI 582", "SI 588", "SI 622"},                       # added shared prereqs
    "LAKES": {"SI 510", "SI 547", "SI 666", "SI 667",
              "SI 676", "SI 580", "SI 647"}
}

# Universal exclusions: prereqs, internship, capstone
UNIVERSAL_EXCLUDE = {"SI 501", "SI 505", "SI 506",
                     "SI 681", "SI 690", "SI 699"}
```

```python
results = {}
for track in ["BDA", "UX", "UCAD", "LAKES"]:
    exclude = UNIVERSAL_EXCLUDE | REQUIRED[track]
    n_students = (track_counts["Inferred_Track"] == track).sum()

    track_df = df_with_tracks[
        (df_with_tracks["Inferred_Track"] == track) &
        (~df_with_tracks["Course_Code"].isin(exclude))
    ]

    elective_counts = (
        track_df.groupby(["Course_Code", "Crse_Descr"])["Emplid"]
        .nunique()
        .sort_values(ascending=False)
        .rename("Student_Count")
        .reset_index()
    )

    elective_counts["Pct_of_Track"] = (
        elective_counts["Student_Count"] / n_students * 100
    ).round(1)

    results[track] = elective_counts

    print(f"\n── {track} (n={n_students} students) ──────────────────")
    print(elective_counts.head(10).to_string(index=False))
```

```python
outside_results = {}

for track in ["BDA", "UX", "UCAD", "LAKES"]:
    exclude = UNIVERSAL_EXCLUDE | REQUIRED[track]
    n_students = (track_counts["Inferred_Track"] == track).sum()

    outside_df = df_with_tracks[
        (df_with_tracks["Inferred_Track"] == track) &
        (~df_with_tracks["Course_Code"].isin(exclude)) &
        (df_with_tracks["Outside_UMSI"] == "Yes")
    ]

    outside_counts = (
        outside_df.groupby(["Course_Code", "Crse_Descr"])["Emplid"]
        .nunique()
        .sort_values(ascending=False)
        .rename("Student_Count")
        .reset_index()
    )
    outside_counts["Pct_of_Track"] = (
        outside_counts["Student_Count"] / n_students * 100
    ).round(1)

    outside_results[track] = outside_counts
    print(f"\n── {track} (n={n_students} students) ──────────────────")
    print(outside_counts.head(10).to_string(index=False))
```
