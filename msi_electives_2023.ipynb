**Define Required Courses by Track**

```python
df_filtered = df[~df["Emplid"].isin(exclude_students)].copy()
print(f"\nFiltered student count: {df_filtered['Emplid'].nunique()}")
```

```python
track_required_courses = {
    "BDA":   ["SI 507", "SI 544", "SI 618", "SI 671"],
    "UX":    ["SI 520"],                                         # only UX-unique
    "UCAD":  ["SI 579", "SI 664", "SI 669"],                    # SI 579 is shared
    "LAKES": ["SI 510", "SI 547", "SI 666", "SI 667",
              "SI 676", "SI 580", "SI 647"]
}
```

```python
# Create flag columns FIRST
for track, courses in track_required_courses.items():
    df_filtered[f"{track}_Flag"] = df_filtered["Course_Code"].isin(courses).astype(int)

```

```python
# Then aggregate
track_counts = df_filtered.groupby("Emplid").agg(
    BDA_Count=("BDA_Flag", "sum"),
    UX_Count=("UX_Flag", "sum"),
    LAKES_Count=("LAKES_Flag", "sum"),
    UCAD_Count=("UCAD_Flag", "sum")
)

track_counts.head()
```

```python
zero_track_students = track_counts[
    (track_counts["BDA_Count"] == 0) &
    (track_counts["UX_Count"] == 0) &
    (track_counts["LAKES_Count"] == 0) &
    (track_counts["UCAD_Count"] == 0)
]
print(zero_track_students)
print(len(zero_track_students), "students have zero required courses for any track.")
```

```python
# Create indicator columns for whether each course is in the required course list.
# Will later be summed per student to count how many required courses they took.
for track, courses in track_required_courses.items():
    df_filtered[f"{track}_flag"] = df_filtered["Course_Code"].isin(courses)
```

```python
# Use the counts of required courses ("votes") to infer the most likely track.
# Calculate total votes, detect ties, and identify the track with the highest votes.
vote_cols = ["BDA_Count", "UX_Count", "LAKES_Count", "UCAD_Count"]
```

```python
# total votes + max vote per student
total_votes = track_counts[vote_cols].sum(axis=1)
max_vote = track_counts[vote_cols].max(axis=1)
```

```python
# how many tracks tied for the max
number_at_max = track_counts[vote_cols].eq(max_vote, axis=0).sum(axis=1)
```

```python
# best guess track
best_guess = (
    track_counts[vote_cols]
    .idxmax(axis=1)
    .str.replace("_Count", "", regex=False)
)
```

```python
# Initialize all students as "Unknown", then assign tracks based on vote results.
track_counts["Inferred_Track"] = "Unknown"
 
# 0 votes → Zero-track (Students with no required-course signals are labeled Zero-track)
track_counts.loc[total_votes.eq(0), "Inferred_Track"] = "Zero-track"
 
# If multiple tracks share the highest vote count, label as "Tie".
tie_mask = total_votes.gt(0) & number_at_max.gt(1)
track_counts.loc[tie_mask, "Inferred_Track"] = "Tie"
 
# non-ties → assign best guess
assign_mask = total_votes.gt(0) & number_at_max.eq(1)
track_counts.loc[assign_mask, "Inferred_Track"] = best_guess[assign_mask]
 
print(track_counts["Inferred_Track"].value_counts())
```

```python
tie_ids = track_counts[track_counts["Inferred_Track"] == "Tie"].index
print(track_counts.loc[tie_ids, ["BDA_Count", "UX_Count", "LAKES_Count", "UCAD_Count"]]
      .value_counts())
```

**Resolve Ties**

```python
# Drop point columns if they already exist (makes cell safe to rerun)
point_cols = ["UX_Points", "UCAD_Points", "BDA_Points", "LAKES_Points"]
track_counts = track_counts.drop(columns=[c for c in point_cols if c in track_counts.columns])
```

```python
# Comprehensive tiebreaker point system
UX_TIEBREAK    = ["SI 520", "SI 539", "SI 582", "SI 588", "SI 622"]
UCAD_TIEBREAK  = ["SI 579", "SI 664", "SI 669"]
BDA_TIEBREAK   = ["SI 608", "SI 630", "SI 649", "SI 650", "SI 670",
                  "SI 568"]   # SI 568 added: required BDA prereq per degree requirements
LAKES_TIEBREAK = ["SI 623", "SI 530", "SI 626", "SI 632", "SI 633", "SI 638",
                  "SI 583", "SI 643", "SI 552", "SI 519"]
                  # SI 626 added: listed as alternative to SI 530 for LAKES per degree requirements
 
df_filtered["UX_Points"]    = df_filtered["Course_Code"].isin(UX_TIEBREAK).astype(int)
df_filtered["UCAD_Points"]  = df_filtered["Course_Code"].isin(UCAD_TIEBREAK).astype(int)
df_filtered["BDA_Points"]   = df_filtered["Course_Code"].isin(BDA_TIEBREAK).astype(int)
df_filtered["LAKES_Points"] = df_filtered["Course_Code"].isin(LAKES_TIEBREAK).astype(int)
 
points = df_filtered.groupby("Emplid")[
    ["UX_Points", "UCAD_Points", "BDA_Points", "LAKES_Points"]
].sum()
 
track_counts = track_counts.join(points, how="left")
track_counts[["UX_Points", "UCAD_Points", "BDA_Points", "LAKES_Points"]] = (
    track_counts[["UX_Points", "UCAD_Points", "BDA_Points", "LAKES_Points"]].fillna(0)
)
```

```python
# Reassign tie students using all four track signals
tie_mask = track_counts["Inferred_Track"].isin(["Tie", "Manual"])
point_cols_map = ["BDA_Points", "UX_Points", "LAKES_Points", "UCAD_Points"]
col_to_track = {
    "BDA_Points":   "BDA",
    "UX_Points":    "UX",
    "LAKES_Points": "LAKES",
    "UCAD_Points":  "UCAD"
}
 
for idx in track_counts[tie_mask].index:
    row = track_counts.loc[idx, point_cols_map]
    max_pts = row.max()
    if max_pts == 0 or (row == max_pts).sum() > 1:
        track_counts.loc[idx, "Inferred_Track"] = "Tie"
    else:
        track_counts.loc[idx, "Inferred_Track"] = col_to_track[row.idxmax()]
 
print(track_counts["Inferred_Track"].value_counts())
```

```python
# Inspect any remaining unresolvable tie students
tie_ids = track_counts[track_counts["Inferred_Track"] == "Tie"].index
print(f"{len(tie_ids)} unresolvable tie students:\n")
 
for emplid in tie_ids:
    courses = sorted(df_filtered[df_filtered["Emplid"] == emplid]["Course_Code"].unique())
    print(f"{emplid}: {courses}")
    print(track_counts.loc[emplid, ["BDA_Count", "UX_Count", "LAKES_Count", "UCAD_Count",
                                     "BDA_Points", "UX_Points", "LAKES_Points", "UCAD_Points"]])
    print()
```

**Resolve Zero-Track**

```python
# Identify students with zero required-course signals and list all courses they took
zero_ids = track_counts[track_counts["Inferred_Track"] == "Zero-track"].index
for emplid in zero_ids:
    courses = sorted(df_filtered[df_filtered["Emplid"] == emplid]["Course_Code"].unique())
    print(f"{emplid}: {courses}")
    print()
```

```python
# Look at their most common courses
zero_courses = df_filtered[df_filtered["Emplid"].isin(zero_ids)]
print("Most common courses among zero-track students:")
print(zero_courses["Course_Code"].value_counts().head(20))
```

```python
# Soft-vote system to assign zero-track students using a broader course signal set
# Mirrors the tiebreaker lists above (SI 568 added for BDA, SI 626 added for LAKES)
zero_track_signals = {
    "UX":    ["SI 520", "SI 539", "SI 582", "SI 588", "SI 622"],
    "UCAD":  ["SI 579", "SI 664", "SI 669"],
    "BDA":   ["SI 608", "SI 630", "SI 649", "SI 650", "SI 670",
              "SI 568"],    # SI 568 added: required BDA prereq per degree requirements
    "LAKES": ["SI 510", "SI 547", "SI 666", "SI 667", "SI 676",
              "SI 580", "SI 647", "SI 530", "SI 626", "SI 552", "SI 538"]
              # SI 626 added: listed as alternative to SI 530 for LAKES
}

soft_votes = df_filtered[df_filtered["Emplid"].isin(zero_ids)].copy()
for track, courses in zero_track_signals.items():
    soft_votes[f"{track}_soft"] = soft_votes["Course_Code"].isin(courses).astype(int)
 
soft_votes = soft_votes.groupby("Emplid")[
    ["UX_soft", "UCAD_soft", "BDA_soft", "LAKES_soft"]
].sum()
print(soft_votes)
```

```python
# Take a closer look at students with 0 soft signals
zero_soft_ids = soft_votes[(soft_votes == 0).all(axis=1)].index.tolist()
print(f"Students with 0 soft signals: {zero_soft_ids}")
 
for emplid in zero_soft_ids:
    print(f"\n{'='*50}")
    print(f"Student: {emplid}")
    print(f"\nCourses taken:")
    print(sorted(df_filtered[df_filtered["Emplid"] == emplid]["Course_Code"].unique()))
    print(f"\nVote counts:")
    print(track_counts.loc[emplid, ["BDA_Count", "UX_Count", "LAKES_Count", "UCAD_Count",
                                     "BDA_Points", "UX_Points", "LAKES_Points", "UCAD_Points",
                                     "Inferred_Track"]])
```

```python
# Assign zero-track students based on soft votes
soft_col_to_track = {
    "UX_soft": "UX",
    "UCAD_soft": "UCAD",
    "BDA_soft": "BDA",
    "LAKES_soft": "LAKES"
}
 
soft_assignments = {}
for emplid, row in soft_votes.iterrows():
    max_votes = row.max()
    if max_votes == 0 or (row == max_votes).sum() > 1:
        soft_assignments[emplid] = "Manual"
    else:
        soft_assignments[emplid] = soft_col_to_track[row.idxmax()]
```

```python
# --- MANUAL OVERRIDES FOR ZERO-TRACK STUDENTS ---
# Review the students labeled "Manual" above and assign them based on their
# full course history. Update the lines below as needed.
# Example (from 2022): soft_assignments["23497908"] = "UX"
#
soft_assignments["54924837"] = "UCAD"
soft_assignments["62200322"] = "UX"
print(soft_assignments)
```

```python
# Apply soft assignments to track_counts
for emplid, track in soft_assignments.items():
    track_counts.loc[emplid, "Inferred_Track"] = track
```

**Elective Analysis**

```python
# Re-merge finalized track labels into course dataset
track_labels = track_counts.reset_index()[["Emplid", "Inferred_Track"]]
df_with_tracks = df_filtered.merge(track_labels, on="Emplid", how="left")
 
print(df_with_tracks["Inferred_Track"].value_counts())
print(f"Unique students: {df_with_tracks['Emplid'].nunique()}")
```

```python
# Required courses per track + universal exclusions
REQUIRED = {
    "BDA":   {"SI 507", "SI 544", "SI 618", "SI 671", "SI 504", "SI 568"},
    "UX":    {"SI 520", "SI 539", "SI 582", "SI 588", "SI 622"},  # shared with UCAD included
    "UCAD":  {"SI 579", "SI 664", "SI 669", "SI 539",
              "SI 582", "SI 588", "SI 622", "SI 504"},                       # shared with UX included
    "LAKES": {"SI 510", "SI 547", "SI 666", "SI 667",
              "SI 676", "SI 580", "SI 647"}
}
```

```python
# Universal exclusions: prereqs, internship, capstone
UNIVERSAL_EXCLUDE = {"SI 501", "SI 505", "SI 506",
                     "SI 681", "SI 690", "SI 699"}
 
results = {}
for track in ["BDA", "UX", "UCAD", "LAKES"]:
    exclude = UNIVERSAL_EXCLUDE | REQUIRED[track]
    n_students = (track_counts["Inferred_Track"] == track).sum()
 
    track_df = df_with_tracks[
        (df_with_tracks["Inferred_Track"] == track) &
        (~df_with_tracks["Course_Code"].isin(exclude))
    ]
 
    elective_counts = (
        track_df.groupby("Course_Code")["Emplid"]
        .nunique()
        .sort_values(ascending=False)
        .rename("Student_Count")
        .reset_index()
    )
    elective_counts["Pct_of_Track"] = (
        elective_counts["Student_Count"] / n_students * 100
    ).round(1)
 
    results[track] = elective_counts
    print(f"\n── {track} (n={n_students} students) ──────────────────")
    print(elective_counts.head(10).to_string(index=False))
```

```python
outside_results = {}
for track in ["BDA", "UX", "UCAD", "LAKES"]:
    exclude = UNIVERSAL_EXCLUDE | REQUIRED[track]
    n_students = (track_counts["Inferred_Track"] == track).sum()
 
    outside_df = df_with_tracks[
        (df_with_tracks["Inferred_Track"] == track) &
        (~df_with_tracks["Course_Code"].isin(exclude)) &
        (df_with_tracks["Outside_UMSI"] == "Yes")
    ]
 
    outside_counts = (
        outside_df.groupby(["Course_Code", "Crse_Descr"])["Emplid"]
        .nunique()
        .sort_values(ascending=False)
        .rename("Student_Count")
        .reset_index()
    )
    outside_counts["Pct_of_Track"] = (
        outside_counts["Student_Count"] / n_students * 100
    ).round(1)
 
    outside_results[track] = outside_counts
    print(f"\n── {track} (n={n_students} students) ──────────────────")
    print(outside_counts.head(10).to_string(index=False))
```

```python
# Export full elective list for all tracks — for student-facing course directory
import json

full_course_map = {}

for track, df_result in results.items():
    # Only include SI courses (not outside UMSI)
    si_only = df_result[df_result['Course_Code'].str.startswith('SI ')]
    for _, row in si_only.iterrows():
        code = row['Course_Code']
        name = row.get('Crse_Descr') or row.get('Crse Descr') or ''
        if code not in full_course_map:
            full_course_map[code] = {'name': name, 'tracks': []}
        if track not in full_course_map[code]['tracks']:
            full_course_map[code]['tracks'].append(track)

# Print as JavaScript array ready to paste into your HTML file
print('const allCourses = [')
for code in sorted(full_course_map.keys(), key=lambda x: int(x.split()[1])):
    info = full_course_map[code]
    tracks = sorted([t.lower() for t in info['tracks']])
    tracks_str = ', '.join(f"'{t}'" for t in tracks)
    name = info['name'].replace("'", "\\'")
    print(f"  {{ code: '{code}', name: '{name}', tracks: [{tracks_str}] }},")
print('];')
```
