**Define Required Courses by Track**

```python
df_filtered = df[~df["Emplid"].isin(exclude_students)].copy()
print(f"\nFiltered student count: {df_filtered['Emplid'].nunique()}")
```

```python
track_required_courses = {
    "BDA": [
        "SI 507",   # fixed — BDA-unique
        "SI 544",   # fixed — BDA-unique
        "SI 568",   # fixed — BDA-unique
        "SI 602",   # fixed — new FA 2022 BDA req
        "SI 618",   # fixed — BDA-unique
        "SI 670",   # pick-1 alternate
        "SI 671",   # pick-1 alternate
    ],
    "UX": [
        "SI 520",   # only UX-unique fixed req (SI 539/582/588/622 overlap with UCAD)
    ],
    "UCAD": [
        "SI 579",   # fixed — UCAD-unique
        "SI 664",   # pick-1 alternate
        "SI 669",   # pick-1 alternate
    ],
    # LAKES covers CIC, MSC, and Digital Curation — all share SI 699 mastery course
    "LAKES": [
        # Shared across all three LAKES sub-pathways
        "SI 510",   # fixed in CIC, MSC, and Digital Curation
        "SI 580",   # pick-1 across CIC, MSC, and Digital Curation
        "SI 647",   # pick-1 across CIC, MSC, and Digital Curation
        # CIC-unique
        "SI 547",   # fixed in CIC only
        # Digital Curation-unique fixed reqs
        "SI 666",   # fixed in DC; also pick-4 in MSC
        "SI 667",   # fixed in DC; also pick-4 in MSC
        "SI 676",   # fixed in DC only
    ]
}
```

```python
# Create flag columns FIRST
for track, courses in track_required_courses.items():
    df_filtered[f"{track}_Flag"] = df_filtered["Course_Code"].isin(courses).astype(int)
```

```python
# Then aggregate
track_counts = df_filtered.groupby("Emplid").agg(
    BDA_Count=("BDA_Flag", "sum"),
    UX_Count=("UX_Flag", "sum"),
    LAKES_Count=("LAKES_Flag", "sum"),
    UCAD_Count=("UCAD_Flag", "sum")
)
track_counts.head()
```

```python
# Use the counts of required courses ("votes") to infer the most likely track
vote_cols = ["BDA_Count", "UX_Count", "LAKES_Count", "UCAD_Count"]
```

```python
# total votes + max vote per student
total_votes = track_counts[vote_cols].sum(axis=1)
max_vote    = track_counts[vote_cols].max(axis=1)
```

```python
# how many tracks tied for the max
number_at_max = track_counts[vote_cols].eq(max_vote, axis=0).sum(axis=1)
```

```python
# best guess track
best_guess = (
    track_counts[vote_cols]
    .idxmax(axis=1)
    .str.replace("_Count", "", regex=False)
)
```

```python
# Initialize all students as "Unknown", then assign tracks based on vote results
track_counts["Inferred_Track"] = "Unknown"
```

```python
# 0 votes → Zero-track
track_counts.loc[total_votes.eq(0), "Inferred_Track"] = "Zero-track"
```

```python
# Tie
tie_mask = total_votes.gt(0) & number_at_max.gt(1)
track_counts.loc[tie_mask, "Inferred_Track"] = "Tie"
```

```python
# non-ties → assign best guess
assign_mask = total_votes.gt(0) & number_at_max.eq(1)
track_counts.loc[assign_mask, "Inferred_Track"] = best_guess[assign_mask]
```

```python
print(track_counts["Inferred_Track"].value_counts())
 
```

**Resolve Ties**

```python
# Drop point columns if they already exist (makes cell safe to rerun)
point_cols = ["UX_Points", "UCAD_Points", "BDA_Points", "LAKES_Points"]
track_counts = track_counts.drop(columns=[c for c in point_cols if c in track_counts.columns])
```

```python
# UX — all fixed reqs + full pick-2 selective list per FA 2022 UX PDF
UX_TIEBREAK = [
    "SI 520",  # fixed
    "SI 539",  # fixed
    "SI 582",  # fixed
    "SI 588",  # fixed
    "SI 622",  # fixed
    "SI 529",  # pick-2 selective
    "SI 552",  # pick-2 selective
    "SI 559",  # pick-2 selective
    "SI 612",  # pick-2 selective
    "SI 616",  # pick-2 selective
    "SI 658",  # pick-2 selective
    "SI 659",  # pick-2 selective
    "SI 684",  # pick-2 selective
]
 
# UCAD — UCAD-distinguishing courses only (SI 504/539/582/588/622 overlap with UX/other)
UCAD_TIEBREAK = [
    "SI 579",  # fixed — UCAD-unique
    "SI 664",  # pick-1 alternate
    "SI 669",  # pick-1 alternate
]
 
# BDA — pick-2 options + pick-1 options + BDA-distinguishing fixed reqs
# SI 561 included as NLP alternative to SI 630 per FA 2022 BDA PDF
BDA_TIEBREAK = [
    "SI 608",  # pick-2 selective
    "SI 630",  # pick-2 selective
    "SI 649",  # pick-2 selective
    "SI 650",  # pick-2 selective
    "SI 670",  # pick-1 alternate
    "SI 671",  # pick-1 alternate
    "SI 568",  # fixed
    "SI 602",  # fixed — new FA 2022 req
    "SI 561",  # NLP alternative to SI 630 per FA 2022 BDA PDF
]
 
# LAKES — covers all three sub-pathways: CIC, MSC, and Digital Curation
# Excludes courses that overlap with UX (SI 539, SI 622) or BDA (SI 649)
# SI 638 REMOVED — not in any FA 2022 PDF
# SI 530 REMOVED — FA 2021 only (SI 626 already covers management signal)
# SI 639 ADDED — Digital Curation pick-2 and MSC recommended elective
# SI 581 ADDED — recommended elective in both CIC and MSC
# SI 549 ADDED — recommended elective in CIC
LAKES_TIEBREAK = [
    # Shared across CIC, MSC, and/or Digital Curation
    "SI 623",  # fixed in CIC and MSC
    "SI 626",  # pick-3 in CIC; pick-4 in MSC
    "SI 632",  # pick-3 in CIC; pick-4 in MSC
    "SI 633",  # pick-3 in CIC; pick-4 in MSC; pick-2 in Digital Curation
    "SI 585",  # pick-4 in MSC; pick-2 in Digital Curation; recommended in CIC
    "SI 519",  # recommended elective in CIC, MSC, and Digital Curation
    "SI 552",  # recommended in CIC and MSC; pick-2 selective in UX (weak LAKES signal)
    "SI 581",  # recommended elective in CIC and MSC
    # CIC-specific
    "SI 547",  # fixed in CIC only (also in primary signals)
    "SI 538",  # pick-3 in CIC only
    "SI 643",  # pick-3 in CIC only
    "SI 549",  # recommended elective in CIC only
    # MSC-specific
    "SI 583",  # pick-4 in MSC only
    # Digital Curation-specific
    "SI 639",  # pick-2 in Digital Curation (paired with SI 564); recommended in MSC
]
```

```python
df_filtered["UX_Points"]    = df_filtered["Course_Code"].isin(UX_TIEBREAK).astype(int)
df_filtered["UCAD_Points"]  = df_filtered["Course_Code"].isin(UCAD_TIEBREAK).astype(int)
df_filtered["BDA_Points"]   = df_filtered["Course_Code"].isin(BDA_TIEBREAK).astype(int)
df_filtered["LAKES_Points"] = df_filtered["Course_Code"].isin(LAKES_TIEBREAK).astype(int)
```

```python
points = df_filtered.groupby("Emplid")[
    ["UX_Points", "UCAD_Points", "BDA_Points", "LAKES_Points"]
].sum()
 
track_counts = track_counts.join(points, how="left")
track_counts[["UX_Points", "UCAD_Points", "BDA_Points", "LAKES_Points"]] = (
    track_counts[["UX_Points", "UCAD_Points", "BDA_Points", "LAKES_Points"]].fillna(0)
)
```

```python
# Reassign tie students using all four track signals
tie_mask = track_counts["Inferred_Track"].isin(["Tie", "Manual"])
point_cols_map = ["BDA_Points", "UX_Points", "LAKES_Points", "UCAD_Points"]
col_to_track = {
    "BDA_Points":   "BDA",
    "UX_Points":    "UX",
    "LAKES_Points": "LAKES",
    "UCAD_Points":  "UCAD"
}
```

```python
for idx in track_counts[tie_mask].index:
    row = track_counts.loc[idx, point_cols_map]
    max_pts = row.max()
    if max_pts == 0 or (row == max_pts).sum() > 1:
        track_counts.loc[idx, "Inferred_Track"] = "Tie"
    else:
        track_counts.loc[idx, "Inferred_Track"] = col_to_track[row.idxmax()]
 
print(track_counts["Inferred_Track"].value_counts())
 
```

```python
tie_ids = track_counts[track_counts["Inferred_Track"] == "Tie"].index
for emplid in tie_ids:
    courses = sorted(df_filtered[df_filtered["Emplid"] == emplid]["Course_Code"].unique())
    print(f"\nStudent: {emplid}")
    print(f"Courses: {courses}")
    print(track_counts.loc[emplid, ["BDA_Count", "UX_Count", "LAKES_Count", "UCAD_Count",
                                     "BDA_Points", "UX_Points", "LAKES_Points", "UCAD_Points"]])
```

```python
# MANUAL OVERRIDES FOR TIE STUDENTS
tie_overrides = {
    # "EMPLID": "TRACK",   # reason: ...
    "21257670": "UX",    # took SI 520, SI 559, SI 616 — UX selectives outweigh CIC
    "59253647": "UX", # took SI 547, SI 643, SI 623 — clear CIC pattern
}

for emplid, track in tie_overrides.items():
    track_counts.loc[emplid, "Inferred_Track"] = track

# Confirm ties are resolved before moving on
remaining_ties = track_counts[track_counts["Inferred_Track"] == "Tie"]
print(f"Remaining ties after manual override: {len(remaining_ties)}")
assert len(remaining_ties) == 0, \
    f"Still have unresolved ties: {list(remaining_ties.index)} — fill in tie_overrides above!"
```

**Resolve Zero-Track**

```python
# Soft-vote signal lists — mirrors primary + tiebreak logic above
zero_track_signals = {
    "UX": [
        "SI 520",  # fixed
        "SI 539",  # fixed
        "SI 582",  # fixed
        "SI 588",  # fixed
        "SI 622",  # fixed
        "SI 529",  # pick-2 selective
        "SI 552",  # pick-2 selective
        "SI 559",  # pick-2 selective
        "SI 612",  # pick-2 selective
        "SI 616",  # pick-2 selective
        "SI 658",  # pick-2 selective
        "SI 659",  # pick-2 selective
        "SI 684",  # pick-2 selective
    ],
    "UCAD": [
        "SI 579",  # fixed — UCAD-unique
        "SI 664",  # pick-1 alternate
        "SI 669",  # pick-1 alternate
    ],
    "BDA": [
        "SI 608",  # pick-2 selective
        "SI 630",  # pick-2 selective
        "SI 649",  # pick-2 selective
        "SI 650",  # pick-2 selective
        "SI 670",  # pick-1 alternate
        "SI 671",  # pick-1 alternate
        "SI 568",  # fixed
        "SI 602",  # fixed — new FA 2022 req
        "SI 561",  # NLP alternative to SI 630 (consistent with BDA_TIEBREAK)
    ],
    # LAKES covers all three sub-pathways: CIC, MSC, and Digital Curation
    "LAKES": [
        # Shared across all three LAKES sub-pathways
        "SI 510",  # fixed in CIC, MSC, and Digital Curation
        "SI 580",  # pick-1 across CIC, MSC, and Digital Curation
        "SI 647",  # pick-1 across CIC, MSC, and Digital Curation
        "SI 623",  # fixed in CIC and MSC
        "SI 626",  # pick-3 in CIC; pick-4 in MSC
        "SI 632",  # pick-3 in CIC; pick-4 in MSC
        "SI 633",  # pick-3 in CIC; pick-4 in MSC; pick-2 in Digital Curation
        "SI 585",  # pick-4 in MSC; pick-2 in Digital Curation; recommended in CIC
        "SI 519",  # recommended in CIC, MSC, and Digital Curation
        "SI 552",  # recommended in CIC and MSC
        "SI 581",  # recommended in CIC and MSC
        # CIC-specific
        "SI 547",  # fixed in CIC only
        "SI 538",  # pick-3 in CIC only
        "SI 643",  # pick-3 in CIC only
        "SI 549",  # recommended in CIC only
        # MSC-specific
        "SI 583",  # pick-4 in MSC only
        # Digital Curation-specific
        "SI 666",  # fixed in DC; also pick-4 in MSC (also in primary signals)
        "SI 667",  # fixed in DC; also pick-4 in MSC (also in primary signals)
        "SI 676",  # fixed in DC only (also in primary signals)
        "SI 639",  # pick-2 in Digital Curation; recommended in MSC
    ]
}
```

```python
soft_votes = df_filtered[df_filtered["Emplid"].isin(zero_ids)].copy()
for track, courses in zero_track_signals.items():
    soft_votes[f"{track}_soft"] = soft_votes["Course_Code"].isin(courses).astype(int)
 
soft_votes = soft_votes.groupby("Emplid")[
    ["UX_soft", "UCAD_soft", "BDA_soft", "LAKES_soft"]
].sum()
 
print(soft_votes)
```

```python
# Check for any students with 0 soft signals (need manual review)
zero_soft_ids = soft_votes[(soft_votes == 0).all(axis=1)].index.tolist()
print(f"Students with 0 soft signals: {zero_soft_ids}")
```

**Final Check**


```python
# Build soft assignments for zero-track students
soft_assignments = {}

for emplid, row in soft_votes.iterrows():
    scores = {
        'BDA': row['BDA_soft'],
        'UX': row['UX_soft'],
        'UCAD': row['UCAD_soft'],
        'LAKES': row['LAKES_soft']
    }
    max_score = max(scores.values())
    if max_score > 0:
        top_tracks = [t for t, s in scores.items() if s == max_score]
        if len(top_tracks) == 1:
            soft_assignments[emplid] = top_tracks[0]
        # ties with equal soft scores left unresolved — handle manually
```

```python
for emplid, track in soft_assignments.items():
    track_counts.loc[emplid, "Inferred_Track"] = track
    
print(track_counts["Inferred_Track"].value_counts())
 
assert "Zero-track" not in track_counts["Inferred_Track"].values, \
    "Still have zero-track students — check soft_assignments!"
 
remaining_ties = track_counts[track_counts["Inferred_Track"] == "Tie"]
if len(remaining_ties) > 0:
    print(f"\nWARNING: {len(remaining_ties)} unresolved tie student(s). "
          "Add a manual override in the section above.")
    for emplid in remaining_ties.index:
        courses = sorted(df_filtered[df_filtered["Emplid"] == emplid]["Course_Code"].unique())
        print(f"  {emplid}: {courses}")
        print(f"  {track_counts.loc[emplid, ['BDA_Count','UX_Count','LAKES_Count','UCAD_Count','BDA_Points','UX_Points','LAKES_Points','UCAD_Points']]}\n")
 
```

```python
REQUIRED = {
    "BDA": {
        "SI 504",   # fixed
        "SI 507",   # fixed
        "SI 544",   # fixed
        "SI 568",   # fixed
        "SI 602",   # fixed (new FA 2022 req)
        "SI 618",   # fixed
        "SI 670",   # pick-1 option
        "SI 671",   # pick-1 option
        "SI 608",   # pick-2 option
        "SI 630",   # pick-2 option
        "SI 649",   # pick-2 option
        "SI 650",   # pick-2 option
        "SI 561",   # pick-2 NLP alt to SI 630
    },
    "UX": {
        "SI 520",   # fixed
        "SI 539",   # fixed
        "SI 582",   # fixed
        "SI 588",   # fixed
        "SI 622",   # fixed
        "SI 529",   # pick-2 option
        "SI 552",   # pick-2 option
        "SI 559",   # pick-2 option
        "SI 612",   # pick-2 option
        "SI 616",   # pick-2 option
        "SI 658",   # pick-2 option
        "SI 659",   # pick-2 option
        "SI 684",   # pick-2 option
    },
    "UCAD": {
        "SI 504",   # fixed
        "SI 539",   # fixed
        "SI 579",   # fixed
        "SI 582",   # fixed
        "SI 588",   # fixed
        "SI 622",   # fixed
        "SI 664",   # pick-1 option
        "SI 669",   # pick-1 option
    },
    # LAKES covers CIC, MSC, and Digital Curation — all pick options included
    "LAKES": {
        # Shared fixed
        "SI 510",   # fixed in CIC, MSC, Digital Curation
        "SI 580",   # pick-1 across all three
        "SI 647",   # pick-1 across all three
        # CIC fixed + picks
        "SI 547",   # fixed in CIC
        "SI 623",   # fixed in CIC and MSC
        "SI 538",   # pick-3 in CIC
        "SI 626",   # pick-3 in CIC; pick-4 in MSC
        "SI 632",   # pick-3 in CIC; pick-4 in MSC
        "SI 633",   # pick-3 in CIC; pick-4 in MSC; pick-2 in DC
        "SI 643",   # pick-3 in CIC
        # MSC-specific picks
        "SI 583",   # pick-4 in MSC
        "SI 585",   # pick-4 in MSC; pick-2 in DC
        "SI 666",   # pick-4 in MSC; fixed in DC
        "SI 667",   # pick-4 in MSC; fixed in DC
        # Digital Curation fixed + picks
        "SI 504",   # fixed in DC (shared with BDA/UCAD)
        "SI 676",   # fixed in DC
        "SI 539",   # pick-2 in DC (shared with UX/UCAD)
        "SI 564",   # pick-2 in DC (paired with SI 639)
        "SI 639",   # pick-2 in DC (paired with SI 564)
        "SI 622",   # pick-2 in DC (shared with UX)
    }
}
```

```python
UNIVERSAL_EXCLUDE = {
    "SI 501",   # core — Contextual Inquiry
    "SI 505",   # remedial/bridge course
    "SI 506",   # core — Programming I
    "SI 681",   # internship
    "SI 690",   # internship
    "SI 699",   # mastery/capstone
}
```

```python
# SI Elective Analysis
results = {}
for track in ["BDA", "UX", "UCAD", "LAKES"]:
    exclude    = UNIVERSAL_EXCLUDE | REQUIRED[track]
    n_students = (track_counts["Inferred_Track"] == track).sum()
 
    track_df = df_with_tracks[
        (df_with_tracks["Inferred_Track"] == track) &
        (~df_with_tracks["Course_Code"].isin(exclude))
    ]
 
    elective_counts = (
        track_df.groupby("Course_Code")["Emplid"]
        .nunique()
        .sort_values(ascending=False)
        .rename("Student_Count")
        .reset_index()
    )
 
    elective_counts["Pct_of_Track"] = (
        elective_counts["Student_Count"] / n_students * 100
    ).round(1)
 
    results[track] = elective_counts
 
    print(f"\n── {track} (n={n_students} students) ──────────────────")
    print(elective_counts.head(10).to_string(index=False))
 
```

```python
# Outside UMSI Electives
outside_results = {}
for track in ["BDA", "UX", "UCAD", "LAKES"]:
    exclude    = UNIVERSAL_EXCLUDE | REQUIRED[track]
    n_students = (track_counts["Inferred_Track"] == track).sum()
 
    outside_df = df_with_tracks[
        (df_with_tracks["Inferred_Track"] == track) &
        (~df_with_tracks["Course_Code"].isin(exclude)) &
        (df_with_tracks["Outside_UMSI"] == "Yes")
    ]
 
    outside_counts = (
        outside_df.groupby(["Course_Code", "Crse_Descr"])["Emplid"]
        .nunique()
        .sort_values(ascending=False)
        .rename("Student_Count")
        .reset_index()
    )
 
    outside_counts["Pct_of_Track"] = (
        outside_counts["Student_Count"] / n_students * 100
    ).round(1)
 
    outside_results[track] = outside_counts
 
    print(f"\n── {track} (n={n_students} students) ──────────────────")
    print(outside_counts.head(11).to_string(index=False))
```
